```python
# ATTENTION! The following exercise is a bit more difficult
# Try to solve it by yourself and if you get stuck check the solution in the notebooks or online.

# Check which season is it based on the month variable.
# If the month is between 3 and 5 print "Spring"
# If the month is between 6 and 8 print "Summer"
# If the month is between 9 and 11 print "Autumn"
# If the month is between 12 and 2 print "Winter"

if month >= 3 and month <= 5:
    print("Spring")
elif month >= 6 and month <= 8:
    print("Summer")
elif month >= 9 and month <= 11:    
    print("Autumn")
else:
    print("Winter")
```
