# 5. For Loops
Let's bring back our list of recordings:
`neuron_activity = [-70, -68.2, -50.5, -39.8, -40, -61.2, -70.2]`

How can we print all the voltages of the neuron? 
From what we learned so far, we would have to manually use the index for each one:

````python
neuron_activity = [-70, -68.2, -50.5, -39.8, -40, -61.2, -70.2]

print(neuron_activity[0])
print(neuron_activity[1])
print(neuron_activity[2])
# ... this is going to take forever if we have 10,000 recordings!
````

A `for` loop solves this exact problem. It allows us to tell the computer: "Take this container of data, and do the same thing to every single item inside it."

```
for <individual_item> in <collection_of_items>:
    <do something>
```
⚠️ As for the conditional statements, notice the spaces before < do something >!

````python
# Let's print every single voltage one by one, the easy way:
for voltage in neuron_activity:
    print(f"Recorded voltage: {voltage} mV")
````

````python
# If we print the variable 'voltage' now, what will its value be?
# It keeps the value of the very last item it processed!
print(f"The loop finished. The last voltage was: {voltage}")
````

**What can we loop over?**
We can loop over "collections" of things (like Lists or Strings). 
We CANNOT loop over a single, isolated number. Let's see what happens if we try:

````python
for i in 5:
    print(i)
````

````python
# We can also loop over text (strings), because a string is just a collection of characters.
word = 'Spike'
for letter in word:
    print(letter)
````

Accumulator: Loops aren't just for printing; they are for calculating! 

We can set up an empty variable BEFORE the loop, and update it INSIDE the loop.

````python
# Exercise: how can we calculate the length of 'neuron_activity' using a for loop?

list_length = 0

for voltage in neuron_activity:
    # To complete: what should we do to list_length every time the loop iterates?
    list_length = list_length + 1

print(f"Total number of recordings calculated manually: {list_length}")
````
