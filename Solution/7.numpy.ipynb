<h1><b>Let's start with NumPy</b></h1> 

https://numpy.org/

NumPy is a library specifically built to easily work with N-dimensional arrays

Libraries are containers for many functions, but they are not included in Python! So the first thing you'll need to do is to include the library in your code!

````python
# Let's import the NumPy library

# Basic import
import numpy

help(numpy)
````

 Whenever you want to use a function contained in the library you need to use a specific sytax: ```library.name_of_the_function```

 To make the syntax as shorter as possible, it is usual to import libraries with an alias (or nickname)

````python
# Let's import numpy with an alias
import numpy as np
````

````python
# Let's create a numpy array from a list
neuron_activity = [-70, -68.2, - 50.5, -39.8, -40, -61.2, -70.2]

neuron_activity_array = np.array(neuron_activity)

# What is the difference between a list and a numpy array?
print(type(neuron_activity)) 
print(type(neuron_activity_array)) 
````

Differences in using arrays

````python
# Check the data type of the elements in the numpy array
print("Elements in the numpy array have data type:", neuron_activity_array.dtype)

# Check the data type of the elements in the list
print("First element of the list has data type:", type(neuron_activity[0]))
print("Second element of the numpy array has data type:", type(neuron_activity_array[1]))
````

````python
# Check the shape, size and dimensions of the numpy array
print("Shape of the numpy array:", neuron_activity_array.shape)
print("Size of the numpy array:", neuron_activity_array.size)
print("Number of dimensions of the numpy array:", neuron_activity_array.ndim)

print("Length of the list:", len(neuron_activity))
print("Length of the numpy array:", len(neuron_activity_array))
print("First element of the list:", neuron_activity[0])
print("First element of the numpy array:", neuron_activity_array[0])
````

<h1>Basic operations for N-dim arrays</h1>

````python
neuron1_activity = np.array([-70, -68.2, - 50.5, -39.8, -40, -61.2, -70.2])
neuron2_activity = np.array([-70.1, -69.7, - 65.3, -50, -40, -64.1, -65.3])
neuron3_activity = np.array([-43.2, -54.7, - 50.5, -36.5, -70, -67.2, -50.2])

# Create an array containing the activity of 3 neurons
neurons_activity_array = np.array([neuron1_activity, neuron2_activity, neuron3_activity])
print(neurons_activity_array.shape)
````

````python
# Sum the activity of the 3 neurons at each time point
total_activity = neuron1_activity + neuron2_activity + neuron3_activity 
print(total_activity)

# Find mean activity of the 3 neurons at each time point
mean_activity = total_activity / 3
print(mean_activity)
````

<h1>Save your data and load it</h1>

````python
# Save the neuronal activity of the 3 neurons in a text file
np.savetxt("neurons_activity.txt", neurons_activity_array)

# Save the neuronal activity of the 3 neurons in a binary file
#np.save("neurons_activity.npy", neurons_activity_array)
````

````python
# Load the neuronal activity of the 3 neurons from the text file
loaded_neurons_activity = np.loadtxt("neurons_activity.txt")
print(loaded_neurons_activity.shape)

# Load the neuronal activity of the 3 neurons from the binary file
loaded_neurons_activity = np.load("neurons_activity.npy")
print(loaded_neurons_activity.shape)
````
