# 1.1 Variables

A variable is a container that stores information. They allow us to store values in memory and to access them later.

The operator `=` assigns information to a variable. For example:

```python
nb_of_spikes = 30
time_window = 10 # seconds
```

The variable `time_window` contains the recording duration (10 seconds) and the variable `nb_of_spikes` contains the number of spikes (30) recorded in that window of time. 

It is possible to display the value of a variable using the function `print`:

```python
print(time_window)
```

```python
print(nb_of_spikes)
```

```python
print(f"{nb_of_spikes = }")
print(f"Variable nb_of_spikes = {nb_of_spikes}")
```

```python
print(f"The neuron fired {nb_of_spikes} times during {time_window} seconds")
```

What happens if we assign a new value to one of these variables?

For example, let's change nb_of_spikes to 40.

```python
nb_of_spikes = 40
print(f"The neuron fired {nb_of_spikes} times during {time_window} seconds")
```

Now we can give the instruction to sum two variables together and store the result in a 3rd variable:
