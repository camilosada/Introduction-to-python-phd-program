# 2. Data types
Data types represent the nature or category of data stored in a variable. The type of data determines what operations we can perform on it.

Basic data types:

* Integers (referred to as `int` in Python): Whole numbers. Examples: 0, -6, 50
* Floats (referred to as `float` in Python): Numbers with a decimal point. Examples: 0.5, 3.14, 1e10
* Strings (referred to as `str` in Python): Text or characters wrapped in quotes. Examples: 'Hello', "Hippocampus", '5'
* Booleans (referred to as `bool` in Python): Represents truth. It can only be True or False.
