# 2. Data types
Data types represent the nature or category of information. The type of data **determines what operations we can perform on it**. It tells the computer how to understand and use the data, like wheather is numbers, text or logical values.

* Integers (referred to as `int` in Python): Whole numbers. Examples: 0, -6, 50
* Floats (referred to as `float` in Python): Numbers with a decimal point. Examples: 0.5, 3.14, 1e10
* Strings (referred to as `str` in Python): Text or characters wrapped in quotes. Examples: 'Hello', "Hippocampus", '5'
* Booleans (referred to as `bool` in Python): Represents truth. It can only be True or False.

Let's create one of each:

```python
nb_of_neurons = 5                # Integer (int)
membrane_potential = -70.5       # Float (float)
brain_region = "Hippocampus"     # String (str)
is_firing = False                 # Boolean (bool)
```

### Indexation
Strings are sequences of characters. We can extract specific elements from a sequence using brackets `[]`. 

⚠️ CAUTION: Python starts counting at 0, not 1!

```python
brain_region = "Hippocampus"

# Get the very first letter
first_letter = brain_region[0]
print(f"The first letter is: {first_letter}")
```

```python
# Get the second letter
second_letter = brain_region[1]
print(f"The second letter is: {second_letter}")
```

```python
# Get the last letter using negative indexing!
last_letter = brain_region[-1]
print(f"The last letter is: {last_letter}")
```
