# 3. Conditional statements
## 3.1 if/else statement
* What if we want to perform an opertion only `if` the neuron is firing?

An if statement checks a condition. If that condition is True, it runs the code inside it. If it is False, it skips it.

```python
if [condition]:
    [do this]
```
⚠️ Python uses empty space (indentation) to know what code belongs inside the if statement. Notice the spaces before [do this]!

````python
is_firing = False  # Try with True

if is_firing:
    print('Neuron is firing')
````

* What if we want to do one thing if the condition is True, and a completely different thing if it is False? 
```python
if [condition]:
    [do this]
else:
    [do this]
```

````python
is_firing = False  # Try with True

if is_firing:
    print('Neuron is firing')
else:
    print("Neuron is not firing")
````

## 3.3 Multiple choices with elif (Else-If)
Sometimes a situation has more than two possible outcomes. We can chain conditions together using `elif`. Python will check them one by one until it finds a True statement, and then it skips the rest.

```python
if [condition]:
    [do this]
elif [condition]:
    [do this]
else:
    [do this]
```
