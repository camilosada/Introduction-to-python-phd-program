# 4. Data Structures

A data structure is a specific way of organizing, storing, and managing data in a computer so that it can be accessed and used efficiently. Think of them as containers.

Python has three basic data structures:

* **Lists (list):** Ordered sequences of data that you can change. 
  Example: `[-70, -68.2, -50.5]` (A sequence of voltages)
* **Tuples (tuple):** Ordered sequences that you CANNOT change once created. 
  Example: `(1024, 768)` (Screen resolution for a visual stimulus)
* **Dictionaries (dict):** Data stored as Key-Value pairs. 
  Example: `{'neuron_id': 'N1', 'is_control': True, 'voltages': [-70, -68.2, -50.5]}`

### Lists
A list is an ordered collection of items, wrapped in square brackets `[]`.

Remember how we used indexing [0] and slicing [0:3] on strings? 
It works the exact same way here!

````python
neuron_activity = [-70, -68.2, - 50.5, -39.8, -40, -61.2, -70.2]
````

````python
# Get the very first recording (Index 0)
first_recording = neuron_activity[0]
print(f"First recording: {first_recording} mV")
````

````python
# Slice the first three recordings (Indices 0, 1, and 2)
window = neuron_activity[0:3]
print(f"First three recordings: {window}")
````

**Useful List Operations**

We can add a new item to the end of our list using `.append()`.
We can also check how many recordings we have in total using `len()`.

````python
# Add a new measurement of -71.5 to the end of the list
neuron_activity.append(-71.5)
print(f"Updated list: {neuron_activity}")
````

````python
# Check the length of the list (how many data points do we have?)
number_of_recordings = len(neuron_activity)
print(f"We have {number_of_recordings} total recordings.")
````

### 4.2 Dictionaries
Instead of using a numbered index (like 0, 1, 2) to find information, 
Dictionaries use labels, which we call "keys". 

They are wrapped in `{}` and use a `key: value` format.

```
{key1: value1, key2: value2}
```

````python
# Let's create a dictionary with information about one recording.
# Notice that we can even put our entire 'neuron_activity' list inside it!
recording_info = {
    "neuron_id": "N1",
    "brain_region": "Hippocampus",
    "is_control": True,
    "voltages": neuron_activity 
}
print(recording_info)
````

````python
# Instead of asking for index [0], we ask for the label (key) we want:
print(f"Neuron ID: {recording_info['neuron_id']}")
print(f"Is this a control cell? {recording_info['is_control']}")
````

````python
# We can even access our entire list of voltages through the dictionary!
print(f"Recorded voltages: {recording_info['voltages']}")
````

**Adding or updating information**

Just like we can append to a list, we can add new labels to a dictionary.

````python
# Let's say we just broke into the cell, but we don't know what type it is yet.
recording_info["cell_type"] = "Unknown"

print("Updated recording info:")
print(recording_info)
````
