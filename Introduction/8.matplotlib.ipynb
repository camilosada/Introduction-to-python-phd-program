```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load the pre-processed data
neurons_activity = np.loadtxt("neurons_activity.txt")
N_neurons, N_time_points = neurons_activity.shape

time_points = range(N_time_points)
print("We have recorded the activity of", N_neurons, "neurons at", N_time_points, "time points.")
```

```python
# Initialize the figure
plt.figure()

plt.plot(neurons_activity[0], label="Neuron 1")
plt.scatter(time_points, neurons_activity[0])

plt.xlabel("Time (ms)")
plt.ylabel("Membrane potential (mV)")

plt.title("Activity of the first over time")

# Save the figure
# plt.savefig("neurons_activity.png")

# Show the figure
plt.show()
```

```python
# Multiple subplots
fig, ax = plt.subplots(1, N_neurons, figsize=(15, 5))

for i in range(N_neurons):
    ax[i].plot(neurons_activity[i], label="Neuron " + str(i+1))
    ax[i].set_xlabel("Time (ms)")
    ax[i].set_ylabel("Membrane potential (mV)")
    ax[i].set_title("Activity of neuron " + str(i+1))
```

```python
# Visualize the distribution of the activity of the 3 neurons at each time point
plt.figure()
for i in range(N_neurons):
    plt.hist(neurons_activity[i], bins=10, label="Neuron " + str(i+1))

plt.legend()
```

```python
# Heatmap of the activity of the 3 neurons at each time point
plt.imshow(neurons_activity, cmap = "viridis")
plt.colorbar(label="Membrane potential (mV)")
plt.xlabel("Time (ms)")
plt.ylabel("Neuron")
```
