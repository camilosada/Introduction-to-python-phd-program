### 6.2 Writing our first function

What if what we need is not implemented in Python? We need to create our own function!

To create a function, we use the `def` keyword (short for "define").
Then we give it a name, tell it what ingredients (inputs) it needs, and tell it what to `return` (output).

```
def <function_name>(<inputs>):
    
    <instructions>
    
    return <output> 
```

Before writing any code, always design your function by answering these three fundamental questions:
1. What will this function be used for? $\rightarrow$ This determines a clear, descriptive function name.
2. What information does the function need to do its job? $\rightarrow$ This identifies your inputs (also called arguments or parameters).
3. Does the function give information back? Which? $\rightarrow$ This defines your returned value (the output).
