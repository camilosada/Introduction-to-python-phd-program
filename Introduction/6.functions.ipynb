# 6. Functions
We now know how to write code to analyze our data! 
But what if we recorded from 50 different cells today? Do we need to copy and paste our `for` loops 50 times? 

A function allows us to save a block of code and reuse it whenever we want.

### 6.1 Writing our first function
To create a function, we use the `def` keyword (short for "define").
Then we give it a name, tell it what ingredients (inputs) it needs, and tell it what to `return` (output).

```
def <function_name>(<inputs>)
    
    <instructions>
    
    return <output>
```

````python
# Let's write a simple function to calculate the firing rate of a neuron
def calculate_firing_rate(spikes, duration):
    rate = spikes / duration
    return rate
````

````python
# Now we can use our recipe on different cells!
cell_1_rate = calculate_firing_rate(30, 10)
print(f"Cell 1 firing rate: {cell_1_rate} Hz")

cell_2_rate = calculate_firing_rate(150, 10)
print(f"Cell 2 firing rate: {cell_2_rate} Hz")
````
